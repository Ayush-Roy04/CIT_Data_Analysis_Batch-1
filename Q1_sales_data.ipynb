```python
df["Name"] = df["Name"].fillna("Unknown")
```

```python
df.loc[df["Product"].isnull(), ["Product", "Sales"]]
```

```python
filtered_products = df[(df['Sales'] > 38000) & (df['Sales'] < 40000)]
print(filtered_products)
```

```python
df["Product"] = df["Product"].fillna("Monitor")
```

```python
df.isna().sum()
```

```python
df.loc[df["Region"].isnull(), ["Name", "Region", "Date"]]
```

```python
df["Region"] = df["Region"].fillna("Unknown")
```

```python
df.loc[df["Sales"].isnull(), ["Product", "Sales"]]
```

```python
keyboard_products = df[df['Product'] == 'Keyboard'][['Product', 'Sales']]
print(keyboard_products)
```

```python
average_keyboard_price = round(keyboard_products['Sales'].mean(), 2)
print(f"The average price of keyboards is: {average_keyboard_price}")
```

```python
df.loc[(df['Product'] == 'Keyboard') & (df['Sales'].isnull()), 'Sales'] = average_keyboard_price
print(df[df['Product'] == 'Keyboard'])
```

```python
printer_products = df[df['Product'] == 'Printer'][['Product', 'Sales']]
print(printer_products)
```

```python
average_printer_price = round(printer_products['Sales'].mean(), 2)
print(f"The average price of printer is: {average_printer_price}")
```

```python
df.loc[(df['Product'] == 'Printer') & (df['Sales'].isnull()), 'Sales'] = average_printer_price
print(df[df['Product'] == 'Printer'])
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.isna().sum()
```

```python
df["Date"] = df["Date"].fillna("0000-00-00")
```

```python
df.to_csv("cleaned_sales.xlsx", index=False)
```
