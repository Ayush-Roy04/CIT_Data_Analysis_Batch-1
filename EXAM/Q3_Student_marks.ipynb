```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("students_marks.csv")
df.shape
```

```python
print(df.head())
print("..........................................................................................")
print(df.tail())
```

```python
numeric_columns = df.select_dtypes(include='number')
print(numeric_columns)
```

```python
subjects = ["English", "Math", "Science", "Arts", "Computer"]
```

```python
#HISTOGRAM

for subject in subjects:
    plt.figure(figsize=(8, 5))

    plt.hist(
        df[subject].dropna(),
        bins=10,
        edgecolor="black",
        alpha=0.7
    )

    plt.title(f"Distribution of {subject} Marks")
    plt.xlabel(f"{subject} Marks")
    plt.ylabel("Number of Students")
    plt.legend([subject])

    plt.show()
```

```python
#Box plot
for subject in subjects:
    plt.figure(figsize=(8, 5))

    plt.boxplot(
        df[subject].dropna(),
        vert=False
    )

    plt.title(f"Boxplot of {subject} Marks")
    plt.xlabel(f"{subject} Marks")
    plt.ylabel("Subject")

    plt.show()
```

```python
#HEATMAP
correlation = df[subjects].corr()

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5,
    fmt=".2f"
)

plt.title("Correlation Matrix of All Subjects")
plt.xlabel("Subjects")
plt.ylabel("Subjects")

plt.show()
```
