```python
import pandas as pd
import numpy as np

customers = pd.read_csv("customers (1).csv")
print(f"Customers shape: {customers.shape}")

orders = pd.read_csv("orders.csv")
print(f"orders shape: {orders.shape}")
```

```python
customers.head()
```

```python
orders.head()
```

```python
merged_df = pd.merge(
    orders,
    customers,
    on="customer_id",
    how="inner")
```

```python
print("Merged DataFrame:")
print(merged_df.head())
```

```python
region_summary = (
    merged_df
    .groupby("region")["order_amount"]
    .agg(
        total_order_amount="sum",
        average_order_amount="mean"
    )
    .reset_index()
)
print(region_summary)
```
